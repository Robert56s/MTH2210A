# Exercice 2

```python
# a)
def u(x):
    return 1/(1 + 25*x**2)
```

```python
# b)
# Interpolation de Lagrange pour n = 6, 8, 10 avec n+1 points équidistants dans [-1,1]

x_plot = np.linspace(-1, 1, 1000)
u_plot = u(x_plot)

degrees = [6, 8, 10]
colors = ['r', 'g', 'b']
plt.figure(figsize=(8,5))
plt.plot(x_plot, u_plot, 'k', label='u(x)')
for n, c in zip(degrees, colors):
    xi = np.linspace(-1, 1, n+1)
    yi = u(xi)
    y_e, lag = MTH2210.lagrange(xi, yi, x_plot)
    plt.plot(x_plot, y_e, c, label=f'n={n}')

plt.legend()
plt.xlabel('x')
plt.ylabel('u(x) et polynômes d\'interpolation')
plt.title("Interpolation de Lagrange — n=6,8,10")
plt.grid(True)
plt.show()
```

```python
# c)

degrees = [6, 8, 10]
colors = ['r', 'g', 'b']
plt.figure(figsize=(8,5))
plt.plot(x_plot, u_plot, 'k', label='u(x)')
for n, c in zip(degrees, colors):
    xi = np.array([ cos(((2*i+1)*pi)/(2*(n+1))) for i in range(0, n+1)])
    print(xi)
    yi = u(xi)
    y_e, lag = MTH2210.lagrange(xi, yi, x_plot)
    plt.plot(x_plot, y_e, c, label=f'n={n}')

plt.legend()
plt.xlabel('x de Chebyshev')
plt.ylabel('u(x) et polynômes d\'interpolation')
plt.title("Interpolation de Lagrange — n=6,8,10")
plt.grid(True)
plt.show()
```

```python
# d)
x_plot = np.linspace(-1, 1, 1000)
u_plot = u(x_plot)

degrees = [6, 8, 10]
colors = ['r', 'g', 'b']
plt.figure(figsize=(8,5))
plt.plot(x_plot, u_plot, 'k', label='u(x)')
for n, c in zip(degrees, colors):
    xi = np.linspace(-1, 1, n+1)
    yi = u(xi)
    y_e, lag = MTH2210.spline_cub(xi, yi, x_plot)
    plt.plot(x_plot, y_e, c, label=f'n={n}')

plt.legend()
plt.xlabel('x')
plt.ylabel('u(x) et polynômes d\'interpolation')
plt.title("Spline cubique naturelle — n=6,8,10")
plt.grid(True)
plt.show()
```
