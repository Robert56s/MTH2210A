```python
# a)

t = np.array([0, 5, 10, 15, 20, 25, 30, 35, 40, 45])
v = np.array([55, 60, 58, 54, 55, 60, 54, 57, 52, 49])


plt.plot(t, v, 'o')
plt.xlabel('t')
plt.ylabel('v')
plt.grid(True)
plt.title("Interpolation polynomiale : Approche de Lagrange pour la vitesse en fonction du temps d'un vehicule")

# ii)
x_plot = np.linspace(0, 45, 1000)
y_e, lag = MTH2210.lagrange(t, v, x_plot)
plt.plot(x_plot, y_e)
plt.show()

#b) 
print(f"v at t=2.5 : {lag(2.5)}")
print(f"v at t=42.5 : {lag(42.5)}")



```

```python
#c)
#i)
def F(tprime):
    total = 1
    for val in t:
        total *= (tprime - val)
    return total

#i)
tprime = np.linspace(0, 45, 1000)
plt.plot(tprime, abs(F(tprime)))
plt.xlabel('tprime')
plt.ylabel('abs(F(tprime))')
plt.grid(True)
plt.title("F(t) dans l'expression analytique de l'errur d'interpolation du polynome de degre 9")
plt.show()
# print(F(np.array([1,2,3])))
```
