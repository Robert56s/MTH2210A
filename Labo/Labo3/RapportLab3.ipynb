# Exercice 1

```python
# Définition d'une fonction équivalente au "gallery('lehmer', size)" de MatLab

def lehmer(size):
    A = np.zeros((size,size))
    for i in range(size):
        for j in range(size):
            A[i,j] = min(i+1,j+1)/max(i+1,j+1)
    return A
```

```python
# a)

A = lehmer(1000)

x = np.ones((1000, 1))

b = np.sum(A, axis=1, keepdims=True)

xAprox = solve(A,b)

E = norm((x - xAprox), 2) / norm(x, 2)

res_norm = norm((b - A@xAprox), 2)

tableau = np.column_stack([E, res_norm])
# Partie b)
print("On remarque que l'erreur relative a deux ordres de grandeur de plus que la norme du residu.\n cela s'explique par l'inégalité du cours suivante : Erreur relative < conditionnement(A)*Résidu_relatif.\n la valeur du Conditionnement(A) vaut environ 1074848  >> 1 . Il s'agit d'un mauvais conditionnement.")
print(" Erreur relative          norme du residu")
print("-"*60)
np.savetxt(sys.stdout, tableau, fmt='%16.15e %16.15e')

print("Ici on remaque que l'erreur relative a six ordres de grandeurs de plus que\n le résidu relatif. Cela s'explique par 'le Conditionnement(A) en vertu de l'inégalité du cours sus-mentionnée.\n En effet, on a bien :")
print(f"\nVoici le residue  relatif : {res_norm/norm(b, 2)}")
print(f"Voici le conditionement de A : {cond(A,2)}")

# print(xAprox)
# print(x)
# print(np.array2string(A, formatter={'float_kind':lambda x: f"{x:.3f}"}, separator=' '))
```
