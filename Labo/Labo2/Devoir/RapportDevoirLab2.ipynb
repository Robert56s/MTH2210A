```python
# Réinitialisation complète de l'environnement
%reset -f

# Import des bibliothèques requises
import numpy as np
import matplotlib.pyplot as plt
import sys
```

# Exercice 

```python
# a)
r1 = 0.99996

n = np.arange(1,11)
r = [r1]

q1 = 1
q2 = 0.33332
q = [q1, q2]

s = np.array([1/(3**(i-1)) for i in n])

# for i in n:
#     r.append(1/3*r[i-1])
#     q.append(10/3*q[i] - q[i-1])

for i in range(1,10):
    r.append(1/3*r[i-1])
for i in range(1,9):
    q.append(10/3*q[i] - q[i-1])

r = np.array(r)
q = np.array(q)

rAbs = abs(s-r)
qAbs = abs(s-q)

tableau = np.column_stack([n,r,q])

print("Tableau 1.1 : suite sn et ses aproximations\n")
print(" n           rn                  qn")
print("-"*60)
np.savetxt(sys.stdout, tableau, fmt='%2d %16.15e %16.15e')



# 3. Créer et configurer le graphe
plt.figure(figsize=(8, 5))
plt.plot(n, rAbs, label='erreur abs(sn-rn)', color='blue', linewidth=2)
plt.plot(n, qAbs, label='erreur abs(sn-qn)', color='red', linewidth=2)

# 4. Ajouter les labels, le titre et la grille pour un rendu scientifique
plt.xlabel('n')
plt.yscale('log')
plt.ylabel("Erreur abs des aproximation de sn")
plt.title('Erreur abs des aproximation de sn en fonction de n')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()

# 5. Afficher le graphe
plt.show()
```

```python
# b)
rPropAbs = rAbs[1:] / rAbs[:-1]
qPropAbs = qAbs[2:] /( qAbs[1:-1] + qAbs[:-2] )

print(rPropAbs)
print(qPropAbs)


```

En observant le graphique et les resultats nous remarquons que le coefficient de la pente de l'erreur absolue de rn est inferieur a 1. Ainsi l'erreur diminue quand n augmente. c'est l'inverse pour l'erreur absolue de qn qui augmente aussi de maniere linéaire sur l'échelle logarithmique. ceci est facilement reperable grace a l'expression approximative des formule 1 et 2.
